```python
# Install Flask and pyngrok, then start the UI with a public URL
!pip install -q flask pyngrok

from pyngrok import ngrok
import threading
import subprocess
import time

def run_flask():
    subprocess.run(["python3", "ui/app.py"], cwd=".", capture_output=True)

thread = threading.Thread(target=run_flask, daemon=True)
thread.start()
time.sleep(3)  # Wait for Flask to start

public_url = ngrok.connect(5001)
print(f"\n✅ Open this URL in your browser to use the Web UI:\n{public_url}")
print("\n• Use the 'Run Hybrid' button to run Hybrid PageRank on Colab's GPU.")
print("• The UI includes Evaluation Metrics (Time, Speedup, Efficiency, Scalability).")
```
